```
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, count, when
import matplotlib.pyplot as plt
```

```
spark = SparkSession.builder.appName('Netflix Big Data Analysis').getOrCreate()
print('Spark Session Created Successfully')
```

```
df = spark.read.csv('netflix_titles.csv', header=True, inferSchema=True)
df.show(10)
```

```
print('Total Rows:', df.count())
print('Total Columns:', len(df.columns))
df.printSchema()
df.describe().show()
```

```
missing_values = df.select([count(when(col(c).isNull(), c)).alias(c) for c in df.columns])
missing_values.show()
```

```
movies_tv = df.groupBy('type').count()
movies_tv.show()

df.groupBy('rating').count().orderBy('count', ascending=False).show()
df.groupBy('release_year').count().orderBy('release_year').show()
df.groupBy('country').count().orderBy('count', ascending=False).show(10)
```

```
type_counts = movies_tv.toPandas()

plt.figure(figsize=(6,4))
plt.bar(type_counts['type'], type_counts['count'])
plt.title('Movies vs TV Shows on Netflix')
plt.xlabel('Content Type')
plt.ylabel('Count')
plt.show()
```
